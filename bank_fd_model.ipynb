```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression

# Ensure Start_Year is present in df_sorted
if 'Start_Year' not in df_sorted.columns:
    df_sorted['Start_Year'] = df_sorted['Financial_Year'].apply(
        lambda x: int(str(x).split('-')[0])
    )

# 1. Print available unique banks
available_banks = sorted(list(df_sorted['Bank_Name'].unique()))
print("Available Banks:")
for idx, bank in enumerate(available_banks, 1):
    print(f"{idx:2d}. {bank}")

# 2. Collect inputs
target_bank = input("\nEnter Bank Name (exact name from the list above): ").strip()
annual_investment = float(input("Enter Annual Investment Amount (in INR): "))
user_age = int(input("Enter your Age: "))
target_goal = float(input("Enter your Target Amount (in INR): "))
simulation_years = int(input("In how many years do you want to achieve this?: "))


rate_col = '5_Year_FD_Rate_Pct'

# Filter data for target bank
bank_data = df_sorted[df_sorted['Bank_Name'].str.lower() == target_bank.lower()].dropna(subset=[rate_col])

if bank_data.empty:
    print(f"\nError: No valid data found for bank '{target_bank}'. Please check the spelling.")
else:
    # Train the predictive model
    X = bank_data[['Start_Year']].values
    y = bank_data[rate_col].values

    model = LinearRegression()
    model.fit(X, y)

    latest_year = int(bank_data['Start_Year'].max())
    current_balance = 0.0

    print("\n" + "="*70)
    print(f" {simulation_years}-YEAR INVESTMENT SIMULATION FOR {target_bank.upper()}")
    print("="*70)

    # Calculation
    for year_idx in range(1, simulation_years + 1):
        future_year = latest_year + year_idx

        # Predict rate
        pred_rate = float(model.predict(np.array([[future_year]]))[0])
        pred_rate = max(0.0, pred_rate)

        # Add extra spread if user qualifies as a senior citizen (typically age >= 60)
        if user_age + year_idx - 1 >= 60:
            pred_rate += 0.50
            rate_type_label = "Senior Rate"
        else:
            rate_type_label = "Standard Rate"


        current_balance += annual_investment

        # Calculate interest earned on current balance
        interest_earned = current_balance * (pred_rate / 100.0)
        current_balance += interest_earned

        print(f"Year {year_idx:2d} ({future_year}): Predicted Rate = {pred_rate:.2f}% ({rate_type_label}) | Interest = INR {interest_earned:,.2f} | Balance = INR {current_balance:,.2f}")

    print("="*70)
    print(f"Final Balance after {simulation_years} Years: INR {current_balance:,.2f}")
    print(f"Target Goal: INR {target_goal:,.2f}")

    if current_balance >= target_goal:
        print("\nSTATUS: YES! POSSIBLE to achieve target!")
    else:
        deficit = target_goal - current_balance
        print(f"\nSTATUS: NOT POSSIBLE. Missed the target by INR {deficit:,.2f}")
    print("="*70)
```
